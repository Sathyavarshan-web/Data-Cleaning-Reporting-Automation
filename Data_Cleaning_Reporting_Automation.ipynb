```python
# Initialize clean_df from the original dataframe
clean_df = df.copy()

# Check duplicate rows before removing
print("Duplicate rows before:", clean_df.duplicated().sum())

# Remove duplicate rows
clean_df = clean_df.drop_duplicates()

# Check duplicate rows after removing
print("Duplicate rows after:", clean_df.duplicated().sum())
```

```python
clean_df["Sales"] = clean_df["Sales"].fillna(clean_df["Sales"].mean())
```

```python
clean_df["Quantity"] = clean_df["Quantity"].fillna(clean_df["Quantity"].median())
```

```python
clean_df["City"] = clean_df["City"].fillna("Unknown")
```

```python
clean_df["Date"] = clean_df["Date"].fillna(method="ffill")
```

```python
clean_df["City"] = clean_df["City"].str.strip().str.title()
```

```python
clean_df["Date"] = pd.to_datetime(clean_df["Date"])
```

```python
# Calculate total sales by product
sales_by_product = clean_df.groupby("Product")["Sales"].sum()

# Create bar chart
plt.figure(figsize=(8,5))
sales_by_product.plot(kind="bar")

plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
sales_by_city = clean_df.groupby("City")["Sales"].sum()

plt.figure(figsize=(8,5))
sales_by_city.plot(kind="bar")

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
product_count = clean_df["Product"].value_counts()

plt.figure(figsize=(7,7))
plt.pie(
    product_count,
    labels=product_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(clean_df["Sales"], bins=5)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
from reportlab.platypus import SimpleDocTemplate, Paragraph
from reportlab.lib.styles import getSampleStyleSheet

# Create PDF document
doc = SimpleDocTemplate("Data_Cleaning_Report.pdf")
styles = getSampleStyleSheet()

elements = []

elements.append(Paragraph("<b>Data Cleaning & Reporting Automation</b>", styles["Title"]))
elements.append(Paragraph(" ", styles["Normal"]))

elements.append(Paragraph("<b>Project Summary</b>", styles["Heading2"]))

elements.append(Paragraph(f"Total Rows: {clean_df.shape[0]}", styles["Normal"]))
elements.append(Paragraph(f"Total Columns: {clean_df.shape[1]}", styles["Normal"]))
elements.append(Paragraph(f"Missing Values: {clean_df.isnull().sum().sum()}", styles["Normal"]))
elements.append(Paragraph(f"Duplicate Rows: {clean_df.duplicated().sum()}", styles["Normal"]))
elements.append(Paragraph(f"Total Sales: {clean_df['Sales'].sum():,.2f}", styles["Normal"]))

elements.append(Paragraph(" ", styles["Normal"]))

elements.append(Paragraph("<b>Cleaning Operations Performed</b>", styles["Heading2"]))
elements.append(Paragraph("• Removed duplicate records.", styles["Normal"]))
elements.append(Paragraph("• Filled missing Sales values with the mean.", styles["Normal"]))
elements.append(Paragraph("• Filled missing Quantity values with the median.", styles["Normal"]))
elements.append(Paragraph("• Filled missing City values with 'Unknown'.", styles["Normal"]))
elements.append(Paragraph("• Standardized City names.", styles["Normal"]))
elements.append(Paragraph("• Converted Date column to datetime format.", styles["Normal"]))

doc.build(elements)

print("PDF Report Created Successfully!")
```
